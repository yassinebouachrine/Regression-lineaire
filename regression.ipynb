```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('Salary_Data.csv')
```

```python
X = data['annees_experience']
Y = data['salaire']
```

```python
plt.scatter(X,Y)
plt.ylabel('Salaire')
plt.xlabel('Les années expersiences')
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split 
X = np.array(data['annees_experience'])
Y = np.array(data['salaire'])
data
```

```python
plt.scatter(X, Y, color='green')
plt.ylabel("Salaire")
plt.xlabel("Les années expersiences")
```

```python
xbar = np.mean(data['annees_experience'])
ybar = np.mean(data['salaire'])
print("La moyenne des années d'éxperience est : "+str(round(xbar,2)))
print("La moyenne des salaires est : "+str(round(ybar,2)))
```

```python
varx = np.var(data['annees_experience'])
ecrx = np.std(data['annees_experience'])
print("La variance des années d'éxperience est : "+str(round(varx,2)))
print("L'ecart-type des années d'éxperience est : "+str(round(ecrx,2)))
```

```python
vary = np.var(data['salaire'])
ecry = np.std(data['salaire'])
print("La variance des salaires est : "+str(round(vary,2)))
print("L'ecart-type des salaires est : "+str(round(ecry,2)))
```

```python
# Divisez les données en ensembles d'entraînement et de test
x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size=1/3)
```

```python
X = X.reshape(-1,1)
Y = Y.reshape(-1,1)
regressor = LinearRegression()
regressor.fit(x_train, y_train)
```

```python
y_predictiontrain = regressor.predict(x_train)
print(y_predictiontrain)
print(y_train)
```

```python
y_predictiontest = regressor.predict(x_test)
print(y_predictiontest)
print(y_test)
```

```python
model = LinearRegression()
model.fit(X,Y)

r_sq = model.score(X,Y)
print('coefficient de corrélation r est: ', round(r_sq, 2))
print('équation de la droite de régression est y = mx + p : ')
print('notre m est :', model.intercept_)
print('notre p est: ', model.coef_)
```

```python
y_train_prediction = regressor.predict(x_train)

plt.figure()
plt.scatter(x_train, y_train, color='green', label="True Value")
plt.plot(x_train, y_train_prediction, color='red', label="Prediction")
plt.ylabel("Salaire")
plt.xlabel("Les années expersiences")
plt.title("Prediction Recult of Training Data")
plt.legend()
plt.show()
```

```python
y_test_prediction = regressor.predict(x_test)

plt.figure()
plt.scatter(x_test, y_test, color='blue', label="True Value")
plt.plot(x_test, y_test_prediction, color='red', label="Prediction")
plt.ylabel("Salaire")
plt.xlabel("Les années expersiences")
plt.title("Prediction Recult of Test Data")
plt.legend()
plt.show()
```

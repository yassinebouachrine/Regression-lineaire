```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt



data = pd.read_csv('Salary_Data.csv')

X = data['annees_experience']
Y = data['salaire']

covariance_matrix = np.cov(X, Y)

# La covariance est située à l'intersection des diagonales de la matrice de covariance
covariance = covariance_matrix[0, 1]

print("Matrice de covariance :" )
print(covariance_matrix)
print("\nCovariance entre X et Y :")
print(covariance)

```
